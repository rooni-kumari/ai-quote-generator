Step 3: AI Quote Generator

```python
import google.generativeai as genai

# Use the model name WITHOUT "models/"
model = genai.GenerativeModel("gemini-2.5-flash")

prompt = "Generate 5 short motivational quotes for college students. Each quote should be on a new line."
response = model.generate_content(prompt)

print("✨ Motivational Quotes ✨\n")
print(response.text)
```
